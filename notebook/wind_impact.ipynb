```python
import importlib
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import app.geography

importlib.reload(app.geography)
load_route = app.geography.load_route
route_bearings = app.geography.route_bearings

project_root = Path.cwd().parent if Path.cwd().name == "notebook" else Path.cwd()
road = load_route((project_root / "data" / "road.gpx").read_bytes(), max_distance_km=None)
```

```python
from concurrent.futures import ThreadPoolExecutor

from app.geography import sample_route
from app.services.weather import get_weather

sampled_route = sample_route(road, every_km=5)
sampled_orientations = route_bearings(sampled_route)

with ThreadPoolExecutor(max_workers=4) as executor:
    weather_values = list(
        executor.map(
            lambda point: get_weather(point["lat"], point["lon"]),
            sampled_route[:-1],
        )
    )

route_weather = [
    {**orientation, "weather": weather}
    for orientation, weather in zip(sampled_orientations, weather_values)
]

route_weather
```
